```python
#installing libraries
import numpy as np
import matplotlib.pyplot as plt
import cv2
import os
```

```python
os.listdir('/content/gdrive/MyDrive/pneumonia_dataset')
```

```python
#setting the training data directory
DataDir = '/content/gdrive/MyDrive/pneumonia_dataset/train/'

os.listdir(DataDir)
```

```python
#initiallizing the category classification or also know as the y-variables
categories = ['NORMAL', 'PNEUMONIA']
```

```python
#convert x-rays into grayscale
for i in categories:
  path = os.path.join(DataDir, i)
  for img in os.listdir(path):
    img_array = cv2.imread(os.path.join(path,img), cv2.IMREAD_GRAYSCALE)
    plt.imshow(img_array, cmap='gray')
    plt.show()
    break
  break
```
